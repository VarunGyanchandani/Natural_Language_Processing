```python
import spacy

# Load the English language model in spaCy
nlp = spacy.load("en_core_web_sm")

# Define the running text for Query 1 and Query 2
running_text_query_1 = """
Nagpur is the third-largest city of the Indian state of Maharashtra after Mumbai and Pune.
Known as the "Orange City", Nagpur is the 13th largest city in India by population.
According to an Oxford's Economics report, Nagpur is projected to be the fifth fastest growing city in the world from 2019 to 2035 with an average growth of 8.41%.
It has been proposed as one of the Smart Cities in Maharashtra and is one of the top ten cities in India in Smart City Project execution.
"""

running_text_query_2 = """
Maharashtra has a Gross State Domestic Product (GSDP) of approximately USD 435 billion (2022-23),
which makes it the largest economy in India. Maharashtra's contribution to India's growth is significant.
The state accounts for 12.92% of India's GDP and has a per capita income that is higher than the national average.
"""

# Function to perform NER and extract relevant information for Query 1
def extract_info_query_1(text):
    doc = nlp(text)
    area_ranking_maharashtra = None
    area_ranking_india = None
    future_growth = None

    for ent in doc.ents:
        if "Area" in ent.text and "Maharashtra" in ent.text:
            area_ranking_maharashtra = ent.text
        elif "Area" in ent.text and "India" in ent.text:
            area_ranking_india = ent.text
        elif "future" in ent.text.lower() and "growth" in ent.text.lower():
            future_growth = ent.text

    return area_ranking_maharashtra, area_ranking_india, future_growth

# Function to perform NER and extract relevant information for Query 2
def extract_info_query_2(text):
    doc = nlp(text)
    contribution_to_economy = None
    gdp_2022_23 = None

    for ent in doc.ents:
        if "GSDP" in ent.text:
            gdp_2022_23 = ent.text
        elif "contribution" in ent.text.lower() and "maharashtra" in ent.text.lower():
            contribution_to_economy = ent.text

    return contribution_to_economy, gdp_2022_23

# Extracting information for Query 1
area_ranking_maharashtra_1, area_ranking_india_1, future_growth_1 = extract_info_query_1(running_text_query_1)
print("Query 1 Results:")
print("Area wise ranking of Nagpur in Maharashtra:", area_ranking_maharashtra_1)
print("Area wise ranking of Nagpur in India:", area_ranking_india_1)
print("Future economic growth of Nagpur:", future_growth_1)

# Extracting information for Query 2
contribution_to_economy_2, gdp_2022_23_2 = extract_info_query_2(running_text_query_2)
print("\nQuery 2 Results:")
print("Contribution of Maharashtra in India’s economy:", contribution_to_economy_2)
print("Maharashtra GSDP for 2022-23:", gdp_2022_23_2)
```

```python
import spacy

# Load the English language model in spaCy
nlp = spacy.load("en_core_web_sm")

# Define the running text
running_text = """
Maharashtra has a Gross State Domestic Product (GSDP) of approximately USD 435 billion (2022-23),
which makes it the largest economy in India. Maharashtra's contribution to India's growth is significant.
The state accounts for 12.92% of India's GDP and has a per capita income that is higher than the national average.
"""

# Function to perform NER and categorize entities
def categorize_entities(text):
    doc = nlp(text)
    entities = {"LOCATION": [], "VALUE": [], "ORG": [], "MISC": []}
    for ent in doc.ents:
        if ent.label_ == "GPE" or ent.label_ == "LOC":
            entities["LOCATION"].append(ent.text)
        elif ent.label_ == "MONEY" or ent.label_ == "QUANTITY":
            entities["VALUE"].append(ent.text)
        elif ent.label_ == "ORG":
            entities["ORG"].append(ent.text)
        else:
            entities["MISC"].append(ent.text)
    return entities

# Extract and categorize entities
categorized_entities = categorize_entities(running_text)

# Print categorized entities
print("Categorized Entities:")
for entity_type, entities in categorized_entities.items():
    print(entity_type + ": ", entities)
```
